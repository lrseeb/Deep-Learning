```python
from transformers import pipeline

MODEL_NAME = 'EleutherAI/gpt-neo-1.3B'

pipe = pipeline('text-generation', 
                model=MODEL_NAME, 
                device_map='auto', 
                dtype='auto')
```

Zero shot example:

```python
prompt =    'Write currencies to countries: ' \
            'Poland => '
```

```python
print(pipe(prompt, max_length=25, max_new_tokens=256))
```

One shot prompting example:

```python
prompt =    'Write currencies to countries: ' \
            'Russia => Rubles' \
            'Poland => '
```

```python
print(pipe(prompt, max_length=25, max_new_tokens=256))
```

Few shot prompting example:

```python
prompt =    'Write currencies to countries: ' \
            'Russia => Rubles;' \
            'Germany => Euro;' \
            'Sweeden => Kronor;' \
            'Poland => '
```

```python
print(pipe(prompt, max_length=25, max_new_tokens=256))
```
