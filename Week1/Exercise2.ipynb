Simplified CNN summary:

### 1st step: Dataset

- The dataset is tokenized and vectorized -> only input_ids are taken into account in this example

### 2nd step: Building the Model

- The CNN model's config is a pretrained model's config, which is used to get the embeddings, vocab_size and convolution layer config.

- Correct activation function is choosen based on model's wanted results -> result is always greater than zero.
`ReLU(x) = max(0, x)`

- Next the pooling layer and output layer are defined

- The forward function is used as an "order list", aka how the model is trained

### 3rd step: Model configuration and training config

- The model is configured based on dataset's vocab_size, embedding (which is the size of the token embeddings), filter size (in this case 3 = 3-gram), and number of filters.

#### Training config

- The training of the model requires information like learning rate (how much the weight updates every step), batch size (the dataset's batch size), and maz steps (the maximum steps when training)


### 4th step: accuracy

- The accuracy is calculated by the amount of how many times the model predicts the label correct by using in this case the test dataset

### 5th step: training and evaluating the model

- Training and evaluation uses the transformers trainer 

- Evaluation can be visualized by matplotlib pyplot library
